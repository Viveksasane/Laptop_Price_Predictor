#### Memory Column :

```python
data["Memory"].value_counts()
```

```python
# Convert to string and clean decimals
data["Memory"] = data["Memory"].astype(str).replace("\.0", "", regex=True)

# Convert TB to GB
data["Memory"] = data["Memory"].str.replace("TB", "000")
data["Memory"] = data["Memory"].str.replace("GB", "")

# Split memory into two layers
new = data["Memory"].str.split("+", n=1, expand=True)

data["first"] = new[0].str.strip()
data["second"] = new[1].fillna("0")

# -------- Layer 1 flags --------
data["Layer1HDD"] = data["first"].apply(lambda x: 1 if "HDD" in x else 0)
data["Layer1SSD"] = data["first"].apply(lambda x: 1 if "SSD" in x else 0)
data["Layer1Hybrid"] = data["first"].apply(lambda x: 1 if "Hybrid" in x else 0)
data["Layer1Flash_Storage"] = data["first"].apply(lambda x: 1 if "Flash Storage" in x else 0)

# -------- Layer 2 flags --------
data["Layer2HDD"] = data["second"].apply(lambda x: 1 if "HDD" in x else 0)
data["Layer2SSD"] = data["second"].apply(lambda x: 1 if "SSD" in x else 0)
data["Layer2Hybrid"] = data["second"].apply(lambda x: 1 if "Hybrid" in x else 0)
data["Layer2Flash_Storage"] = data["second"].apply(lambda x: 1 if "Flash Storage" in x else 0)

# Remove non-numeric characters
data["first"] = data["first"].str.replace(r"\D", "", regex=True).astype(int)
data["second"] = data["second"].str.replace(r"\D", "", regex=True).astype(int)

# -------- Final storage columns --------
data["HDD"] = (data["first"] * data["Layer1HDD"]) + (data["second"] * data["Layer2HDD"])
data["SSD"] = (data["first"] * data["Layer1SSD"]) + (data["second"] * data["Layer2SSD"])
data["Hybrid"] = (data["first"] * data["Layer1Hybrid"]) + (data["second"] * data["Layer2Hybrid"])
data["Flash_Storage"] = (
    data["first"] * data["Layer1Flash_Storage"] +
    data["second"] * data["Layer2Flash_Storage"]
)

# Drop helper columns
data.drop(
    [
        "first", "second",
        "Layer1HDD", "Layer1SSD", "Layer1Hybrid", "Layer1Flash_Storage",
        "Layer2HDD", "Layer2SSD", "Layer2Hybrid", "Layer2Flash_Storage"
    ],
    axis=1,
    inplace=True
)
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
def tune_regression_models(X_train, X_test, Y_train, Y_test, models, params, cv=5):

    print("Regression Model Tuning Results")
    print("-" * 60)

    for name, model in models.items():
        print(f"\n🔹 Model: {name}")

        # ✅ If model has NO parameters → just fit
        if name not in params:
            model.fit(X_train, Y_train)
            print("No hyperparameters to tune")

            ypre = model.predict(X_test)
            print(f"Training R² : {model.score(X_train, Y_train):.4f}")
            print(f"Test R²     : {r2_score(Y_test, ypre):.4f}")
            continue

        # ✅ If parameters exist → GridSearch
        grid = GridSearchCV(
            estimator=model,
            param_grid=params[name],
            cv=cv,
            scoring="r2",
            n_jobs=-1
        )

        grid.fit(X_train, Y_train)

        # Best parameters & score from cross-validation
        print(f"Best R² Score : {grid.best_score_:.4f}")
        print(f"Best Params  : {grid.best_params_}")

        # Evaluate best estimator on train and test set
        best_model = grid.best_estimator_
        ypred = best_model.predict(X_test)
        print(f"Training R² : {best_model.score(X_train, Y_train):.4f}")
        print(f"Test R²     : {r2_score(Y_test, ypred):.4f}")
```
